```python
import pandas as pd
import pytz

# Time Zone Mapping (Expanded)
timezone_map = {
    'KS': 'US/Central', 'NC': 'US/Eastern', 'MS': 'US/Central', 'WA': 'US/Pacific',
    'TN': 'US/Central', 'TX': 'US/Central', 'WI': 'US/Central', 'OH': 'US/Eastern',
    'MI': 'US/Eastern', 'GA': 'US/Eastern', 'VA': 'US/Eastern', 'KY': 'US/Eastern',
    'PA': 'US/Eastern', 'WV': 'US/Eastern', 'MO': 'US/Central', 'ID': 'US/Mountain',
    'NE': 'US/Central', 'QC': 'Canada/Eastern', 'NY': 'US/Eastern', 'MN': 'US/Central',
    'OR': 'US/Pacific', 'ON': 'Canada/Eastern', 'IL': 'US/Central', 'IN': 'US/Eastern',
    'MD': 'US/Eastern', 'NJ': 'US/Eastern', 'IA': 'US/Central', 'NH': 'US/Eastern',
    'AL': 'US/Central', 'SC': 'US/Eastern', 'CT': 'US/Eastern', 'FL': 'US/Eastern',
    'OK': 'US/Central', 'LA': 'US/Central', 'AR': 'US/Central', 'DE': 'US/Eastern',
    'AB': 'Canada/Mountain', 'NB': 'Canada/Atlantic', 'SK': 'Canada/Central',
    'MA': 'US/Eastern', 'ME': 'US/Eastern', 'ND': 'US/Central', 'MB': 'Canada/Central',
    'VT': 'US/Eastern', 'SD': 'US/Central', 'RI': 'US/Eastern', 'BC': 'Canada/Pacific',
    'CO': 'US/Mountain', 'NS': 'Canada/Atlantic', 'PE': 'Canada/Atlantic',
    'VI': 'Australia/Victoria', 'TS': 'Australia/Tasmania', 'HI': 'US/Hawaii',
    'AK': 'US/Alaska', 'CA': 'US/Pacific', 'WY': 'US/Mountain', 'MT': 'US/Mountain',
    'AZ': 'US/Mountain', 'UT': 'US/Mountain', 'SA': 'Australia/South',
    'QL': 'Australia/Queensland', 'AC': 'Australia/Sydney', 'NV': 'US/Pacific',
    'NT': 'Australia/Darwin', 'NM': 'US/Mountain', 'DC': 'US/Eastern',
    'NL': 'Canada/Newfoundland', 'YT': 'Canada/Pacific', 'GU': 'Pacific/Guam',
    'NU': 'Canada/Central'
}

# Hemisphere Mapping (Expanded)
hemisphere_map = {
    'KS': 'Northern', 'NC': 'Northern', 'MS': 'Northern', 'WA': 'Northern',
    'TN': 'Northern', 'TX': 'Northern', 'WI': 'Northern', 'OH': 'Northern',
    'MI': 'Northern', 'GA': 'Northern', 'VA': 'Northern', 'KY': 'Northern',
    'PA': 'Northern', 'WV': 'Northern', 'MO': 'Northern', 'ID': 'Northern',
    'NE': 'Northern', 'QC': 'Northern', 'NY': 'Northern', 'MN': 'Northern',
    'OR': 'Northern', 'ON': 'Northern', 'IL': 'Northern', 'IN': 'Northern',
    'MD': 'Northern', 'NJ': 'Northern', 'IA': 'Northern', 'NH': 'Northern',
    'AL': 'Northern', 'SC': 'Northern', 'CT': 'Northern', 'FL': 'Northern',
    'OK': 'Northern', 'LA': 'Northern', 'AR': 'Northern', 'DE': 'Northern',
    'AB': 'Northern', 'NB': 'Northern', 'SK': 'Northern', 'MA': 'Northern',
    'ME': 'Northern', 'ND': 'Northern', 'MB': 'Northern', 'VT': 'Northern',
    'SD': 'Northern', 'RI': 'Northern', 'BC': 'Northern', 'CO': 'Northern',
    'NS': 'Northern', 'PE': 'Northern', 'VI': 'Southern', 'TS': 'Southern',
    'HI': 'Northern', 'AK': 'Northern', 'CA': 'Northern', 'WY': 'Northern',
    'MT': 'Northern', 'AZ': 'Northern', 'UT': 'Northern', 'SA': 'Southern',
    'QL': 'Southern', 'AC': 'Southern', 'NV': 'Northern', 'NT': 'Southern',
    'NM': 'Northern', 'DC': 'Northern', 'NL': 'Northern', 'YT': 'Northern',
    'GU': 'Northern', 'NU': 'Northern'
}

def get_season(month):
    if month in [6, 7, 8]:
        return 'Summer'
    elif month in [9, 10, 11]:
        return 'Autumn'
    elif month in [12, 1, 2]:
        return 'Winter'
    else:
        return 'Spring'

def get_local_month(row):
    date = row['Failure Date']
    state = row['Owner State']
    if pd.isna(date) or pd.isna(state):
        return None
    try:
        tz_str = timezone_map[state]
        tz = pytz.timezone(tz_str)
        local_date = date.tz_convert(tz) if date.tzinfo else date.tz_localize('UTC').tz_convert(tz)
        return local_date.month
    except KeyError:
        print(f"Timezone not found for state: {state}")
        return None

def get_adjusted_season(row):
    month = row['Local Failure Month']
    state = row['Owner State']
    hemisphere = hemisphere_map.get(state)

    if month is None or hemisphere is None:
        return None

    if hemisphere == 'Northern':
        return get_season(month)
    else:
        if month in [12, 1, 2]:
            return 'Summer'
        elif month in [3, 4, 5]:
            return 'Autumn'
        elif month in [6, 7, 8]:
            return 'Winter'
        else:
            return 'Spring'

# Apply the functions to create the 'failure Season' column
claim_cleaned_df['Local Failure Month'] = claim_cleaned_df.apply(get_local_month, axis=1)
claim_cleaned_df['Failure Season'] = claim_cleaned_df.apply(get_adjusted_season, axis=1)

# Drop the Local Failure Month
claim_cleaned_df.drop(columns=['Local Failure Month'], inplace=True)

# Print the first few rows
print(claim_cleaned_df.head())
```

## Text Mining on Textual Columns

```python
# Import necessary libraries

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import nltk
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
import re
```

```python
# Download NLTK resources
nltk.download('punkt')  # For tokenization
nltk.download('stopwords')
nltk.download('wordnet')  # For lemmatization
```

```python
# Text Preprocessing: Defining Components
# removal of stop words
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
```

```python
# Text Preprocessing: Tokenization, Stemming, Stopword Removal, and Cleaning

def preprocess_text(text):
    # convert all the letters to Lowercasing
    text = text.lower()
    print(f"\nAfter Lowercasing: '{text}'")

    # Removing Punctuation and Special Characters from the text
    text = re.sub(r'[^\w\s]', '', text)
    print(f"After Punctuation Removal: '{text}'")

    # Tokenization - Splitting the text into Words (tokens)
    tokens = nltk.word_tokenize(text)
    print(f"After Tokenization: {tokens}")

    # Stopword Removal and Stemming
    cleaned_tokens = []
    for word in tokens:
        if word not in stop_words:
            stemmed_word = stemmer.stem(word)
            cleaned_tokens.append(stemmed_word)
    print(f"After Stopword Removal and Stemming: {cleaned_tokens}")

    # Joining Tokens Back into a String
    return ' '.join(cleaned_tokens)
```

```python
# Applying Text Preprocessing on failure part data columns Failure Part Description Cleaned, Complaint Diagnosis_Cleaned, Work Performed_Cleaned

failure_part_cleaned_df['Failure Part Description Cleaned'] = failure_part_cleaned_df['Failure Part Description'].apply(preprocess_text)
failure_part_cleaned_df['Work Performed_Cleaned'] = failure_part_cleaned_df['Work Performed'].apply(preprocess_text)
failure_part_cleaned_df['Complaint Diagnosis_Cleaned'] = failure_part_cleaned_df['Complaint Diagnosis'].apply(preprocess_text)
```

```python
# apply TF-IDF Vectorization on Work Performed

tfidf_vectorizer_work = TfidfVectorizer(
    stop_words='english',  # Using built-in stop words for TF-IDF
    max_features=500,  # Limiting features
    ngram_range=(1, 2)  # Using unigrams and bigrams
)
work_tfidf_matrix = tfidf_vectorizer_work.fit_transform(failure_part_cleaned_df['Work Performed_Cleaned'])
work_tfidf_features_df = pd.DataFrame(work_tfidf_matrix.toarray(), columns=tfidf_vectorizer_work.get_feature_names_out())
print("\nWork Performed TF-IDF Features:")
print(work_tfidf_features_df.head())
```

```python
# TF-IDF Vectorization: Complaint Diagnosis

tfidf_vectorizer_diagnosis = TfidfVectorizer(
    stop_words='english',
    max_features=500,
    ngram_range=(1, 2)
)
diagnosis_tfidf_matrix = tfidf_vectorizer_diagnosis.fit_transform(failure_part_cleaned_df['Complaint Diagnosis_Cleaned'])
diagnosis_tfidf_features_df = pd.DataFrame(diagnosis_tfidf_matrix.toarray(), columns=tfidf_vectorizer_diagnosis.get_feature_names_out())
print("\nComplaint Diagnosis TF-IDF Features:")
print(diagnosis_tfidf_features_df.head())
```

```python
# TF-IDF Vectorization: Failure Part Description
tfidf_vectorizer_failure_part_descrip = TfidfVectorizer(
    stop_words='english',
    max_features=500,
    ngram_range=(1, 2)
)
failure_part_descrip_tfidf_matrix = tfidf_vectorizer_diagnosis.fit_transform(failure_part_cleaned_df['Failure Part Description Cleaned'])
failure_part_descrip_tfidf_features_df = pd.DataFrame(failure_part_descrip_tfidf_matrix.toarray(), columns=tfidf_vectorizer_diagnosis.get_feature_names_out())
print("\nFailure Part Description TF-IDF Features:")
print(failure_part_descrip_tfidf_features_df.head())
```

```python
# Calculating Row-Wise TF-IDF Sum

failure_part_cleaned_df['tfidf_work_sum'] = work_tfidf_features_df.sum(axis=1)

failure_part_cleaned_df['tfidf_diagnosis_sum'] = diagnosis_tfidf_features_df.sum(axis=1)

failure_part_cleaned_df['tfidf_failure_part_descrip_sum'] = failure_part_descrip_tfidf_features_df.sum(axis=1)
```

```python
# Aggregate failure part data on claim number

# since, we have 2 or more failure parts for some claim numbers, we are will be keeping the first occurence for those claim numbers.

failure_part_aggregated_df = failure_part_cleaned_df.groupby('Claim Number').agg({
    'Vendor Number': 'first',  # Keep the first occurrence of Vendor Number
    'Failure Part Series': 'first',  # Keep the first occurrence of Failure Part Series
    'tfidf_work_sum': 'sum',  # Sum of all tfidf_work_sum values
    'tfidf_diagnosis_sum': 'sum',  # Sum of all tfidf_diagnosis_sum values
    'tfidf_failure_part_descrip_sum': 'sum',  # Sum of all tfidf_failure_part_descrip_sum values
    'work_performed_sentiment': 'sum',  # Sum of all work_sentiment values
    'diagnosis_sentiment': 'sum',  # Sum of all diagnosis_sentiment values
    'Failure Description': 'first',  # Keep the first occurrence of Failure Description
    'Failure Part': 'count'  # Count of Failure Part ids
}).rename(columns={'Failure Part': 'Number of Failure Parts'}).reset_index()
```
