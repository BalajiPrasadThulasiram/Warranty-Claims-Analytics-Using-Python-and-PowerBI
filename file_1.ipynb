```python
# It is incorrectly clasisfied as object, we need to convert to date format
# Also we won't be needing time for analysis, so we will convert to date only instead of datetime
# When we tried to convert, we got an error OutOfBoundsDatetime: Out of bounds nanosecond timestamp: 9998-12-31 00:00:00,
# at position 129, it means that our data has values with years like 9998 which are not valid

# Pandas datetime format only supports dates within the range 1677-09-21 to 2262-04-11.
# Check for the purchase date column to find the invalid dates

import pandas as pd

# Check the unique years in the column
claim_df['Purchase Date'] = claim_df['Purchase Date'].astype(str)  # Ensuring it's a string before extracting years
claim_df['Purchase Year'] = claim_df['Purchase Date'].str[:4]  # Extract only the year

# Convert to numeric and check for anomalies
claim_df['Purchase Year'] = pd.to_numeric(claim_df['Purchase Year'], errors='coerce')
invalid_dates = claim_df[(claim_df['Purchase Year'] < 1900) | (claim_df['Purchase Year'] > 2100)]

print("Rows with invalid purchase dates:")
print(invalid_dates[['Purchase Date']])
```

```python
# Replace invalid dates with NaT
claim_df.loc[(claim_df['Purchase Year'] < 1900) | (claim_df['Purchase Year'] > 2100), 'Purchase Date'] = pd.NaT

#cleaning the purchase date field
claim_df['Purchase Date'] = claim_df['Purchase Date'].astype(str).str.strip()
claim_df['Purchase Date'] = pd.to_datetime(claim_df['Purchase Date'], errors='coerce')
print(claim_df['Purchase Date'].dtype)  # Should now be datetime64[ns]
```

```python
# Removing time component from the date time

claim_df['Purchase Date'] = claim_df['Purchase Date'].dt.date
claim_df['Purchase Date'] = pd.to_datetime(claim_df['Purchase Date']).dt.normalize()
```
